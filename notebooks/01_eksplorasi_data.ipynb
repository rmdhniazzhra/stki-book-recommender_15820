```python
import pandas as pd
import numpy as np

# Baca dataset books.csv
df = pd.read_csv('books.csv')
print(f"Total baris data awal: {len(df)}")
df.head(3)
```

```python
# Ambil atribut teks yang relevan
kolom_stki = ['title', 'authors', 'categories', 'description', 'average_rating']
df_stki = df[kolom_stki].copy()

# Cek jumlah data kosong
print("Jumlah data kosong per kolom:")
print(df_stki.isnull().sum())
```

```python
# Hapus baris yang tidak memiliki deskripsi (karena teks deskripsi adalah inti retrieval)
df_clean = df_stki.dropna(subset=['description']).reset_index(drop=True)
print(f"Total data bersih siap pakai: {len(df_clean)}")

# Contoh isi sinopsis buku pertama
print("\n--- Contoh Data Buku Pertama ---")
print("Judul   :", df_clean.loc[0, 'title'])
print("Kategori:", df_clean.loc[0, 'categories'])
print("Sinopsis:\n", df_clean.loc[0, 'description'])
```

```python
import re

def simple_clean_text(text):
    # 1. Lowercase (case folding)
    text = text.lower()
    # 2. Hapus karakter selain huruf dan spasi
    text = re.sub(r'[^a-z\s]', '', text)
    # 3. Hapus spasi berlebih
    text = ' '.join(text.split())
    return text

# Terapkan ke kolom description
df_clean['cleaned_description'] = df_clean['description'].apply(simple_clean_text)

# Tampilkan perbandingan sebelum dan sesudah cleaning
print("Sebelum Cleaning:\n", df_clean.loc[0, 'description'][:150])
print("\nSesudah Cleaning:\n", df_clean.loc[0, 'cleaned_description'][:150])
```

```python
# Simpan dataset hasil cleaning ke file baru
df_clean.to_csv('cleaned_books.csv', index=False)
print("Dataset bersih berhasil disimpan ke cleaned_books.csv!")
```
